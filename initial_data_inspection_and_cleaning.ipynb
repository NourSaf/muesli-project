```python
shipment_data = orderstable_campaigndata_orderprocessdata_interndata 
```

```python
shipment_data.info()
```

## Deleting unnecessary rows

```python
# Adding the dates

shipment_data.drop(labels=['Customer Name_x', 'Origin Channel', 'City', 'Postal Code','Region', 'Sales','Discount','Profit'],axis=1)
```

## Adding campaign information

```python
# Add order day type: weekday or weekend
shipment_data['Order Date Day Name'] = shipment_data['Order Date_x'].dt.day_name()
```

## Adding Times

```python
# Time between the order is done until its loaded to the truck
shipment_data['Processing Time'] = shipment_data['On Truck Scan Date'] - shipment_data['Order Date_x']
```

```python
# Time between order is loaded to the truck and is delivered to the customer
shipment_data['Delivery Time'] = shipment_data['Arrival Scan Date'] - shipment_data['On Truck Scan Date']
```

```python
#  Time between product is ready to be shipped and it's loaded into the truck
shipment_data['Dispatch time'] = shipment_data['On Truck Scan Date'] - shipment_data['Ready to Ship Date']
```
